# Output Formatting

### The following program prints a neat table
### Format specs control width and alignment: &lt; left, &gt; right, ^ centre, plus a width and decimals, e.g. {price:&gt;8.2f}.

```python
# Print the header Item Price: Item left-aligned in 10 characters, then Price
print(f"{'Item':<10}Price")

# For the items Pen (5) and Notebook (45.5), print each row with the name left-aligned in 10 characters and the price right-aligned in 5 
# characters with 1 decimal: Pen 5.0 and Notebook 45.5
items = [("Pen", 5), ("Notebook", 45.5)]
total = 0.0
for name, price in items:
    print(f"{name:<10}{price:>5.1f}")
    total += price

# Print a line of 15 dashes
print("-" * 15)

# Print Total 50.50: Total left-aligned in 10, the total right-aligned in 5 with 2 decimals
print(f"{'Total':<10}{total:>5.2f}")

# Print **Done** by centring Done in 8 characters filled with *
print(f"{'Done':*^8}")
```
