## Data Types

### int, float, complex, bool, str, list, tuple, range, dictionary, set

```python
# int example : ints are whole numbers irrespective if they are positive or negative
age = 25
count = 10
temp = -10
print(age, count, temp)
print(type(temp))
print(type(temp).__name__)
```

```python
# float example : float represent fractional component 
number = 25.0
print(type(number))
```

```python
# complex contains imaginary component 
number = 2+1j
print (type(number))
```

```python
# Boolean (bool) python has two boolean values True or False 
is_active = True
is_complete = False 
print (type(is_active))
```

```python
# String it is represented as text inside the quotes 
a = "FDE"
b = "-45"
c = "55"
print(type(b))
```

```python
# list is an ordered and mutable collection and allow duplicate values []
my_list = [10, "veera", "-1"] 
print (my_list)
print (type(my_list))

another_list = [20, "test", -1, "test"]
print (another_list)
print (type(another_list))

# we can even nest one list inside another
yet_another_list = [my_list, another_list]
print (yet_another_list)
print (type(yet_another_list))
```

```python
# tuple an ordered collection and immutable ().
my_tuple = (10, "veera", "-1") 
print (my_tuple)
print (type(my_tuple))

another_tuple = (10, 20, 30, 10, "test")
print (another_tuple)
print (type(another_tuple))

# we can nest tuples also
yet_another_tuple = (my_tuple, another_tuple)
print (yet_another_tuple)
print (type(yet_another_tuple))
```

```python
# Set is unordered, mutable collection of unique element {}
number_set = {10, 20, 30, 30}
print (type(number_set))
```

```python
# Dictionary (dict) stores key-values pairs
student = {
    "name": "ravi",
    "age" : 25,
    "course" : "FDE"
}
print(student)
print(type(student))
```

```python
#None Type means absence of the value
result = None 
print(type(result))
```

```python
# range is a sequence of integers 
number = range(5)
print(number)
print (list(number))
print (type(number))
```

## Converison types process 
### converts a value of one data type to another data type 

```python
# float to int converison 
a = 2
b = 3.5
c = a+b
c = int(a+b)
print(c)
print(type(c))
```

```python
# another example of float to int, the decimal is truncated.

num = 3.1
num1 = 3.4
num2 = 3.5
num3 = 3.6
num4 = 3.9
print(int(num), int(num1), int(num2), int(num3), int(num4))
```

```python
# string to integer 
age ="25"
age = int(age)
print(age)
print(type(age))
```

```python
# another example of string to int but the string has floating point value
num = "3.1"
num1 = "3.4"
num2 = "3.5"
num3 = "3.6"
num4 = "3.9"
#print(float(num), float(num1), float(num2), float(num3), float(num4))
print(int(num), int(num1), int(num2), int(num3), int(num4))
```
