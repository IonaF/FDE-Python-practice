### String Slicing 

#### Slicing means extracting a portion of string 
#### start index : where slicing begins, included 
#### stop index : where slicing ends, excluded
#### step : number of positions to shift 

##### Positive Step : start should be less than end (left to right) ---> Incremental slicing  
##### Negative Step : start should be greater than stop (right to left) ---> Decremental slicing 

## Ex: [start :stop : step]

    <tr>
        <td>P</td><td>Y</td><td>T</td><td>H</td><td>O</td><td>N</td>

```python
word = "PYTHON"
print(word[3:5])
#start is 3
#stop is 4
```

```python
word = "PYTHON"
print(word[3:3])
#start is 5
#stop is 3
##no output
```

```python
word = "PYTHON"
print(word[2:5])
#start is 3
#stop is 4
## python slicing does not give any error if stop range is out of the range and also start point larger than end point 
```

```python
# slicing by omitting the start index
word = "PYTHON"
print(word[:3])
print(word[:4])
print(word[:10])
```

```python
# slicing by omitting the stop index
word = "PYTHON"
print(word[3:])
print(word[1:])
print(word[:])
```

```python
# Negative slicing process
word = "PYTHON"
print(word[-4:3])
```

```python
word = "PYTHON"
print(word[-4:2])
```

```python
word = "PYTHON"
print(word[-4:-5:1])
```

```python
word = "PYTHON"
print(word[-4:-5:-1])
```

```python
word = "PYTHON"
print(word[-4:-6:-1])
```
