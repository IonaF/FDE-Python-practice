## Left shift and Right shift 

### Left shift operation

```python
# Left shift example 1
x = 15
x <<= 1
print(x)

# before left shift the binary equivalent of 15 is 1111 ie. 1 x 2^3 + 1 x 2^2 + 1 x 2^1 + 1 x 2^0
# left shift (<<) : move bits to left and add place 0 in the empty position on the right
# ie. 11110.
# converting this back to decimal we have 1 x 2^4 + 1 x 2^3 + 1 x 2^2 + 1 x 2^1 + 0 x 2^0 which is 30
```

```python
# Left shift example 2
x = 19 # 10011 in binary form
x <<= 2 # 1001100 op performing left shift
print (x) # converting result back to decimal
print(2**6 + 2 ** 3 + 2 ** 2)
```

```python
# Left shift example 3
x = 21 # 10101 in binary format
# print(2**4 + 2 ** 2 + 2 ** 0)
x <<= 3 # 10101000
print(x)
# print(2 ** 7 + 2 ** 5 + 2 ** 3)
```

### Right shift operation

```python
# Right shift example 1 
x = 15 # this is 1111 in binary form
x >>= 1 # drop the right-most bit (1 bit) and pad with 1 0 on the left which makes 0111
print(x) # this gives (1 x 2^2 + 1 x 2^1 + 1 x 2^0) which is 7

# >>1 
# before right shift x=15 1111
# after right shift 0111      
```

```python
# Right shift example 2 
x = 15 # this is 1111 in binary form
x >>= 4 # dropping 4 bits on the right we get 0
print (x) 
```

### Summary of left shift and right shift

```python
# Left shift 
15        1111
15 << 1   11110
15 << 2   111100
15 << 3   1111000
```

```python
# Right shift
15        1111
15 >>1    0111
15 >>2    0011
15 >>3    0001
```

## Bitwise Operators (&, |, ^, ~)

### Bitwise AND

### In bitwise AND (&) we are basically multiplying the bits in each position of first no's binary form with corresponding bit in second no's binary form 

```python
# example 1 of Bitwise AND
a = 5 # 0101 in binary form
b = 3 # 0011 in binary form
print ("Bitwise AND:", a & b)
```

```python
# example 2 of Bitwise AND
a = 8 # 1000 in binary form
b = 4 # 0100 in binary form 
print ("Bitwise AND:", a & b) # [1 x 0][0 x 1][0 x 0][0 x 0] which is 0
```

### Bitwise OR

#### For Bitwise OR (|) only if bits in corresponding positions of both binary numbers is 0 then result is 0, else 1

```python
# example of Bitwise OR
a = 5 # 0101 in binary form
b = 3 # 0011 in binary form 
print ("Bitwise OR:", a | b) #[0][1][1][1] which is 7
```

### Bitwise XOR

#### For Bitwise XOR (^) only if either of bits in corresponding positions of both binary numbers is 1 then result is 1, else 0

```python
# example of Bitwise XOR 
a = 5 # 0101 in binary form
b = 3 # 0011 in binary form
print ("Bitwise XOR:", a ^ b) # [0^0][1^0][0^1][1^1] => [0][1][1][0] => 0 x 2 ** 3 + 1 x 2 ** 2 + 1 x 2 ** 1 + 0 x 2 ** 0 => 4 + 2 = > 6
```

### Bitwise Not

#### Reversing the bit in question, short cut is we increment the number by 1 and multiply with -1

```python
#example of Bitwsie not
a = 6 # 0110 in binary form
print ("Bitwise NOT:", ~a)
#~a = -(a+1)
```
