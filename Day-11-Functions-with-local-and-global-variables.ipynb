## Function with Local and Global variables 
### global variable : defined outside the function 
### local variable : define inside the function

```python
def naresh ():
    name = "veera" # inside of the function local variable 
    print ("hello", name)
naresh()
```

```python
name = "FDE"
def naresh ():
    name = "veera" # inside the function local variable 
    print ("hello", name) # print local variable
naresh()
print(name) #print global variable 
```

```python
salary = 50000
def update_salary():
    global salary #this global block tells python -> Dont create a new local varible use existing variable named "salary" 
    salary= 60000
final=update_salary()
print(salary)

#intially salary is = 50000
#global block tells python to use the global variable so its change to 60000
```
